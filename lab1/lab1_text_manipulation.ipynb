# Part 1 – String methods

Python strings come with methods that cover most of the basic cleaning operations. They are the first tools to try before writing a regular expression.

## 1.1 Case conversion

```python
text = "Hello World! Welcome to NLP."

print("lower():     ", text.lower())       # all lowercase
print("upper():     ", text.upper())       # all uppercase
print("title():     ", text.title())       # first letter of each word
print("capitalize():", text.capitalize())  # first character only
```

```python
sentence = "The Quick BROWN Fox Jumps Over The Lazy DOG"

result = sentence.lower()  # YOUR CODE HERE

print(result)
assert result == "the quick brown fox jumps over the lazy dog", "Check your answer"
```

## 1.2 Length and counting

```python
text = "banana"

print("Length:    ", len(text))            # number of characters
print("Count 'a': ", text.count("a"))      # occurrences of a substring
print("Count 'an':", text.count("an"))
```

## 1.3 Whitespace, split and join

```python
text = "   Hello World!   "

print(f"strip():  '{text.strip()}'")     # both ends
print(f"lstrip(): '{text.lstrip()}'")    # left only
print(f"rstrip(): '{text.rstrip()}'")    # right only

text2 = "###Hello###"
print(f"strip('#'): '{text2.strip('#')}'")  # strip specific characters
```

```python
# split() turns a string into a list, join() does the opposite
print("Split by comma:     ", "apple,banana,cherry".split(","))
print("Split by whitespace:", "Hello   World   NLP".split())   # any run of whitespace

words = ["Natural", "Language", "Processing"]
print("Join with space: ", " ".join(words))
print("Join with hyphen:", "-".join(words))
```

```python
messy_text = "   This    text   has    irregular     spacing   "

clean_text = " ".join(messy_text.split())  # YOUR CODE HERE

print(f"Clean: '{clean_text}'")
assert clean_text == "This text has irregular spacing", "Check your answer"
```

## 1.6 Mini-challenge: a basic cleaning function

```python
# Exercise 1.6 – Write `basic_clean(text)` that:
#   1. converts to lowercase
#   2. replaces newlines by spaces
#   3. replaces runs of spaces by a single space
#   4. removes leading/trailing whitespace
def basic_clean(text):
    """Lowercase the text and normalize its whitespace."""
    # 1. convert to lowercase
    text = text.lower()
    # 2. replace newlines by spaces
    text = text.replace('\n', ' ')
    # 3 & 4. split and join to handle runs of spaces and strip leading/trailing
    text = " ".join(text.split())
    return text

test_text = """   HELLO   World!
   This is    a TEST.   """

result = basic_clean(test_text)
print(f"Result: '{result}'")
assert result == "hello world! this is a test.", "Check your function"
```

# Part 2 – Regular expressions

A regular expression (regex) describes a *pattern* of characters. The `re` module uses such patterns to search, extract, split and replace text. The reference is https://docs.python.org/3/library/re.html.

## 2.1 The four main functions

```python
text = "My phone number is 123-456-7890 and my zip code is 12345."

# re.search(): first match (or None). \d+ means "one or more digits"
match = re.search(r"\d+", text)
if match:
    print("search():  ", match.group())

# re.findall(): every match, as a list
print("findall(): ", re.findall(r"\d+", text))

# re.sub(): replace every match
print("sub():     ", re.sub(r"\d", "X", text))

# re.split(): split on a pattern
print("split():   ", re.split(r"[;,\s]+", "apple;banana,cherry orange"))
```

```python
# Exercise 2.1 – Extract every year (4-digit number) from the text.
text = "Python was created in 1991. TensorFlow was released in 2015. GPT-3 came out in 2020."

years = re.findall(r"\d{4}", text)

print(years)

assert years == ["1991", "2015", "2020"], "Check your pattern"
```

## 2.2 Character classes

```python
text = "Hello World! 123 test_var"

print("\\d  (digits):        ", re.findall(r"\d", text))
print("\\w+ (word chars):    ", re.findall(r"\w+", text))     # letters, digits and _
print("\\s  (whitespace):    ", re.findall(r"\s", text))
print("\\S+ (non-whitespace):", re.findall(r"\S+", text))
```

```python
# Custom classes with [...]
text = "The price is $50.99 or €45.00"

print("Vowels:    ", re.findall(r"[aeiouAEIOU]", text))
print("Currency:  ", re.findall(r"[$€£]", text))
print("Non-digits:", re.findall(r"[^0-9]+", text))   # ^ inside [] means "not"
```

```python
text = "Hello world NLP is GREAT for text processing 123"

lowercase_words = re.findall(r"\b[a-z]+\b", text)

print(lowercase_words)
assert lowercase_words == ["world", "is", "for", "text", "processing"], "Check your pattern"
```

## 2.3 Quantifiers

```python
text = "a aa aaa aaaa b bb bbb"

print("a+     (1 or more):", re.findall(r"a+", text))
print("ba*    (0 or more):", re.findall(r"ba*", text))
print("ba?    (0 or 1):   ", re.findall(r"ba?", text))
print("a{2}   (exactly 2):", re.findall(r"a{2}", text))
print("a{2,3} (2 to 3):   ", re.findall(r"a{2,3}", text))
```

```python
# Greedy vs non-greedy
html = "<p>First</p><p>Second</p>"
print("Greedy:    ", re.findall(r"<p>.*</p>", html))    # as much as possible
print("Non-greedy:", re.findall(r"<p>.*?</p>", html))   # as little as possible
```

```python

text = "Call me at 123-456-7890 or 987-654-3210. My old number was 555-1234."

phone_numbers = re.findall(r"\b\d{3}-\d{3}-\d{4}\b", text)

print(phone_numbers)
assert phone_numbers == ["123-456-7890", "987-654-3210"], "Check your pattern"
```

## 2.4 Anchors and word boundaries

```python
text = "Hello World"
print("Starts with 'Hello':", bool(re.match(r"^Hello", text)))   # ^ = start of string
print("Ends with 'World':  ", bool(re.search(r"World$", text)))  # $ = end of string

text2 = "cat catalog caterpillar"
print("All 'cat':       ", re.findall(r"cat", text2))
print("Whole word 'cat':", re.findall(r"\bcat\b", text2))        # \b = word boundary
```

```python

text = "I need to prepare a presentation. The prerequisites are comprehensive."

pre_words = re.findall(r"\bpre\w+", text)  # YOUR CODE HERE

print(pre_words)
assert pre_words == ["prepare", "presentation", "prerequisites"], "Check your pattern"
```

## 2.5 Groups and alternation

```python
# Parentheses capture parts of the match
text = "John Smith: john.smith@email.com, Jane Doe: jane.doe@company.org"
print("Email parts:", re.findall(r"(\w+\.\w+)@(\w+\.\w+)", text))

# | means "or"
text2 = "I have a cat and a dog. My neighbor has a bird."
print("Pets found: ", re.findall(r"cat|dog|bird", text2))
```

```python
# Exercise 2.5 – Extract the dates written as DD/MM/YYYY or DD-MM-YYYY as (day, month, year) tuples.
# Hint: use three groups and a class [/-] for the separator.
text = "Important dates: 25/12/2024, 01-01-2025, and 14/02/2025."

dates = re.findall(r"(\d{2})[/-](\d{2})[/-](\d{4})", text)

print(dates)
assert dates == [("25", "12", "2024"), ("01", "01", "2025"), ("14", "02", "2025")], "Check your pattern"
```

## 2.6 Cleaning patterns used in NLP

```python
sample_text = """
Check out https://example.com for more info!
Contact us at support@company.com 📧
Follow @nlp_expert on Twitter! #MachineLearning #NLP
Price: $99.99 (50% off!!!)
"""

print("No URLs:   ", re.sub(r"https?://\S+", "", sample_text))
print("No emails: ", re.sub(r"\S+@\S+", "[EMAIL]", sample_text))
print("No social: ", re.sub(r"[@#]\w+", "", sample_text))          # mentions and hashtags
```

```python
text = "Hello!!!   What's up???  This is so cool..."

print("No repeated punctuation:", re.sub(r"([!?.]){2,}", r"\1", text))   # \1 = first group
print("No extra whitespace:    ", re.sub(r"\s+", " ", text))
print("Alphanumeric only:      ", re.sub(r"[^\w\s]", "", text))
```

## 2.7 Flags

```python
text = "Python PYTHON python PyThOn"
print("Without flag:   ", re.findall(r"python", text))
print("With IGNORECASE:", re.findall(r"python", text, re.IGNORECASE))

multiline_text = """First line
Second line
Third line"""
# With MULTILINE, ^ and $ also match at line boundaries
print("Capitalized line starts:", re.findall(r"^[A-Z]\w+", multiline_text, re.MULTILINE))
```

## 2.8 Final challenge: a complete text preprocessor

```python
# Exercise 2.8 – Write `preprocess_text(text)` that applies, in this order:
#   1. lowercase
#   2. remove URLs (http:// or https://)
#   3. remove e-mail addresses
#   4. remove mentions (@user) and hashtags (#topic)
#   5. remove punctuation: keep only letters, digits and whitespace
#   6. replace runs of whitespace by a single space
#   7. strip leading/trailing whitespace
def preprocess_text(text):
    """Clean a raw text for NLP tasks."""

    # Step 1: lowercase
    text = text.lower()

    # Step 2: remove URLs
    text = re.sub(r"https?://\S+", "", text)

    # Step 3: remove e-mails
    text = re.sub(r"\S+@\S+", "", text)

    # Step 4: remove mentions and hashtags
    text = re.sub(r"[@#]\w+", "", text)

    # Step 5: remove punctuation
    text = re.sub(r"[^\w\s]", "", text)

    # Step 6: normalize whitespace
    text = re.sub(r"\s+", " ", text)

    # Step 7: strip
    return text.strip()
    pass
raw_text = """
    🚀 Check out our NEW product at https://example.com!!!
    Contact: sales@company.com @CompanyName #Innovation #Tech
    Limited time offer: 50% OFF!!!
"""

cleaned = preprocess_text(raw_text)
print(f"Cleaned: '{cleaned}'")
assert cleaned == "check out our new product at contact limited time offer 50 off", "Check the order of the steps"
```
